# Dataset Parity

Focused examples for bundled ggplot2-compatible datasets.

```python
from ggplotly import (
    aes,
    data,
    geom_line,
    geom_point,
    geom_segment,
    ggplot,
    labs,
    scale_x_date,
    theme_minimal,
)
```

## Economics Date Axis

```python
economics = data("economics")

(ggplot(economics, aes(x="date", y="unemploy"))
 + geom_line(color="steelblue", linewidth=1.5)
 + scale_x_date(date_breaks="10 years", date_labels="%Y")
 + labs(title="US unemployment over time", x="Date", y="Unemployed")
 + theme_minimal())
```

## Presidential Terms

```python
presidential = data("presidential")

(ggplot(
    presidential,
    aes(x="start", xend="end", y="name", yend="name", color="party"),
)
 + geom_segment(linewidth=4)
 + scale_x_date(date_breaks="4 years", date_labels="%Y")
 + labs(title="US presidential terms", x="Term dates", y="President")
 + theme_minimal())
```

## Texas Housing Decimal Year

```python
houston = data("txhousing").loc[lambda df: df["city"] == "Houston"]

(ggplot(houston, aes(x="date", y="sales"))
 + geom_line(color="firebrick", linewidth=1.2)
 + geom_point(color="firebrick", size=2)
 + labs(title="Houston housing sales by decimal year", x="Decimal year", y="Sales")
 + theme_minimal())
```
