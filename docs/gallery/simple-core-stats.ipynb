# Simple Core Stats

Focused examples for ggplot2-compatible core stats and thin geom wrappers.

```python
import sys
from pathlib import Path

import numpy as np
import pandas as pd

repo = Path.cwd()
if not (repo / "ggplotly").exists():
    repo = next(parent for parent in Path.cwd().parents if (parent / "ggplotly").exists())
sys.path.insert(0, str(repo))

from ggplotly import *

rng = np.random.default_rng(404)
points = pd.DataFrame({"x": [0], "y": [0]})
blank_bounds = pd.DataFrame({"x": [-10, 10], "y": [-5, 5]})

area = pd.DataFrame({
    "year": [2020, 2022, 2021, 2022],
    "value": [2, 6, 5, 8],
    "series": ["North", "North", "South", "South"],
})

sample = pd.DataFrame({"x": rng.normal(0, 1, 500)})
scatter = pd.DataFrame({
    "x": np.linspace(0, 10, 160),
})
scatter["y"] = 1.5 + 0.55 * scatter["x"] + rng.normal(0, 0.4 + scatter["x"] * 0.08, len(scatter))

hex_data = pd.DataFrame({
    "x": rng.normal(0, 1.0, 450),
    "y": rng.normal(0, 1.0, 450),
})
hex_data["z"] = 2 + 0.8 * hex_data["x"] - 0.6 * hex_data["y"] + rng.normal(0, 0.25, len(hex_data))

density_input = pd.DataFrame({
    "group": np.repeat(["A", "B"], [180, 260]),
    "value": np.r_[rng.normal(-0.4, 0.55, 180), rng.normal(0.65, 0.75, 260)],
})
ydensity_data, _ = stat_ydensity(mapping={"x": "group", "y": "value"}, n=120, scale="count").compute(density_input)
```

## geom_blank

```python
(ggplot(points, aes(x="x", y="y"))
 + geom_blank(data=blank_bounds, mapping=aes(x="x", y="y"))
 + geom_point(size=8, color="#1f77b4")
 + labs(title="Blank layer trains wider axes", x="x", y="y"))
```

## stat_align

```python
(ggplot(area, aes(x="year", y="value", fill="series"))
 + geom_area(alpha=0.45, position="identity")
 + geom_point(aes(color="series"), size=5)
 + scale_x_continuous(breaks=[2020, 2021, 2022], labels=["2020", "2021", "2022"])
 + labs(title="Area groups align to a shared x domain", x="year", y="value"))
```

## geom_function

```python
(ggplot(sample, aes(x="x"))
 + geom_density(fill="#9ecae1", color="#3182bd", alpha=0.45)
 + geom_function(fun=lambda x: np.exp(-0.5 * x**2) / np.sqrt(2 * np.pi), xlim=(-4, 4), color="#111111", size=2)
 + labs(title="Function overlay on density", x="x", y="density"))
```

## geom_quantile

```python
(ggplot(scatter, aes(x="x", y="y"))
 + geom_point(alpha=0.35, color="#636363", size=4)
 + geom_quantile(quantiles=(0.1, 0.5, 0.9), size=2)
 + labs(title="Quantile regression lines", x="x", y="y"))
```

## stat_summary_hex

```python
(ggplot(hex_data, aes(x="x", y="y", z="z"))
 + stat_summary_hex(bins=16, fun="mean")
 + labs(title="Hex bins colored by mean z", x="x", y="y"))
```

## stat_ydensity

```python
(ggplot(ydensity_data, aes(x="y", y="violinwidth", color="group"))
 + geom_line(size=2)
 + labs(title="Y-density profiles scaled by group count", x="value", y="violin width"))
```
