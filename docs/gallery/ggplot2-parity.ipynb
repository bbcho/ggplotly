# ggplot2 Parity

Focused examples for ggplot2-compatible aliases and thin wrappers.

```python
import sys
from pathlib import Path

import numpy as np
import pandas as pd

repo = Path.cwd()
if not (repo / "ggplotly").exists():
    repo = next(parent for parent in Path.cwd().parents if (parent / "ggplotly").exists())
sys.path.insert(0, str(repo))

from ggplotly import *

rng = np.random.default_rng(303)
cloud = pd.DataFrame({
    "x": np.r_[rng.normal(-1.1, 0.45, 160), rng.normal(1.0, 0.5, 160)],
    "y": np.r_[rng.normal(0.0, 0.55, 160), rng.normal(1.25, 0.45, 160)],
})
```

## geom_bin_2d

```python
(ggplot(cloud, aes(x="x", y="y"))
 + geom_bin_2d(bins=18)
 + scale_fill_binned(low="#f7fbff", high="#08519c", name="count")
 + labs(title="2D binned counts", x="x", y="y"))
```

## geom_density2d

```python
(ggplot(cloud, aes(x="x", y="y"))
 + geom_point(alpha=0.25, size=3, color="#666666")
 + geom_density2d(bins=8, n=60, color="#111111", linewidth=1.2)
 + labs(title="2D density contour lines", x="x", y="y"))
```

## geom_density2d_filled

```python
(ggplot(cloud, aes(x="x", y="y"))
 + geom_density2d_filled(bins=9, n=60, palette="Viridis", alpha=0.85)
 + geom_point(alpha=0.25, size=2, color="#333333")
 + labs(title="Filled 2D density contours", x="x", y="y"))
```
